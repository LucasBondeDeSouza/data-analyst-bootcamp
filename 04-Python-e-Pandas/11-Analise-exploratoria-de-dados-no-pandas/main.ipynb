<h1>Análise Exploratória de dados em pandas</h1>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

<h2>1. Visualizar os Dados</h2>

```python
df = pd.read_csv("world_population.csv")
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

<h2>2. Info()</h2>

```python
df.info()
```

<h2>3. Describe()</h2>

```python
df.describe()
```

<h2>4. Contando todos os valores nulos</h2>

```python
df.isnull().sum()
```

<h2>5. Contando todos os valores únicos</h2>

```python
df.nunique()
```

<h2>6. Classificação por valores</h2>

```python
df.sort_values(by="World Population Percentage", ascending=False).head(10)
```

<h2>7. Correlação entre Colunas</h2>

```python
df.corr(numeric_only=True)
```

<h2>8. Mapa de calor usando Seaborn</h2>

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.rcParams['figure.figsize'] = (20, 7)
plt.show()
```

<h2>9. Visualizando Agrupamento de Dados</h2>

```python
df.groupby('Continent').mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
```

```python
df[df['Continent'].str.contains('Oceania')]
```

```python
df2 = df.groupby('Continent')[['1970 Population', 
    '1980 Population', '1990 Population', '2000 Population', 
    '2010 Population', '2015 Population', '2020 Population', 
    '2022 Population']].mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
df2
```

```python
df3 = df2.transpose()
df3
```

```python
df3.plot()
```
