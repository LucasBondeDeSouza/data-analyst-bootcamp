```python
from bs4 import BeautifulSoup
import requests
import time
import datetime
import csv
import pandas as pd

import smtplib
```

```python
def check_price():
    URL = 'https://www.netshoes.com.br/p/tenis-nike-court-vision-low-next-nature-masculino-branco-2IC-7368-014?campaign=gglepqpla&selected=2IC-7368-014-38&gad_source=1&gad_campaignid=20286482792&gbraid=0AAAAADje7YaL6_tegbQlawz-4nuBWtW1P&gclid=Cj0KCQjw4qHEBhCDARIsALYKFNMVuB5F4gHAB4j5tBw2mxqxUTNtI8rCRsoJV5z_s-l8qwEMjiJl7UQaAp5sEALw_wcB&gclsrc=aw.ds'

    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept-Language": "pt-BR,pt;q=0.9,en-US;q=0.8,en;q=0.7"
    }

    response = requests.get(URL, headers=headers)

    soup = BeautifulSoup(response.content, "html.parser")

    title = soup.find(class_= 'product-name').get_text()
    price = soup.find(class_ = 'saleInCents-value').get_text()

    title = title.strip()
    price = price.strip()[2:].split(',')[0]

    today = datetime.date.today()

    header = ['Title', 'Price', 'Date']
    data = [title, price, today]

    with open('NetshoesScraperDataset.csv', 'a+', newline = '', encoding = 'UTF8') as f:
        writer = csv.writer(f)
        writer.writerow(data)
```

```python
while(True):
    check_price()
    time.sleep(5)
```
